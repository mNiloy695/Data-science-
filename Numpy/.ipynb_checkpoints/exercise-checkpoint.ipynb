```python
import numpy as np
```

```python
data = np.array([
      [18, 85, 78],   # Student 1\n",
      [19, 92, 88],   # Student 2\n",
      [17, 76, 95],   # Student 3\n",
      [18, 65, 70],   # Student 4\n",
      [20, 90, 85]    # Student 5\n",
    ])
```

```python
data
```

```python
#Get the shape of the matrix.
data.shape
```

```python
#Find the average age of students
average_age=np.sum(data[:,:1],axis=0)/5
average_age
```

```python
#Extract Math marks of all students.\n
math_marks_all_student=np.sum(data[:,1:2],axis=0)
math_marks_all_student
```

```python
#Find the highest Science mark
science_mark=data[:,2:]
highest_mark=np.max(science_mark) 
highest_mark
```

```python
#Increase Math marks of all students by 5
new_marks=data.copy()
new_marks=new_marks[:,1:2]+5
new_marks
```

```python
#Find how many students are younger than 19
slice=data[:,:1]
younger_student=slice>19
younger_student
```

```python
np.sum(younger_student)
```

```python
#Get details of the student who scored more than 90 in Math.
slice=data[:,1:2] 
more_than_90=slice>90
data[more_than_90.ravel()]
```

```python
#Calculate the average marks in each subject (column-wise mean)
slice=data[:,1:]
np.sum(slice)/10
```
