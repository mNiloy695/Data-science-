```python
import numpy as np
```

```python
arr = np.random.randint(0, 10, size=(9, 9))
```

```python
arr
```

```python
shape=arr.shape
shape
```

```python
row=shape[0]
col=shape[1]
```

```python
row,col
```

```python
for i in range(0,row-2,3):
    for j in range(0,col-2,3):
        slice=arr[i:i+3,j:j+3]
        print(slice)
```
